```python
import os
import cv2
import numpy as np
from tqdm import tqdm

# =====================================
# 경로
# =====================================

INPUT_DIR = "./test"
OUTPUT_DIR = "./test_crop"
DEBUG_DIR = "./test_debug"

os.makedirs(OUTPUT_DIR, exist_ok=True)
os.makedirs(DEBUG_DIR, exist_ok=True)


# =====================================
# 잎 크롭 함수
# =====================================

def crop_leaf(img):

    original = img.copy()

    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # 배경 제거용 blur
    blur = cv2.GaussianBlur(gray, (5, 5), 0)

    # 자동 threshold
    _, thresh = cv2.threshold(
        blur,
        0,
        255,
        cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU
    )

    # 노이즈 제거
    kernel = np.ones((5,5), np.uint8)

    thresh = cv2.morphologyEx(
        thresh,
        cv2.MORPH_CLOSE,
        kernel,
        iterations=2
    )

    contours, _ = cv2.findContours(
        thresh,
        cv2.RETR_EXTERNAL,
        cv2.CHAIN_APPROX_SIMPLE
    )

    if len(contours) == 0:
        return original, original

    largest = max(contours, key=cv2.contourArea)

    x, y, w, h = cv2.boundingRect(largest)

    # margin 추가
    margin = 15

    x = max(0, x - margin)
    y = max(0, y - margin)

    w = min(img.shape[1] - x, w + margin*2)
    h = min(img.shape[0] - y, h + margin*2)

    crop = original[y:y+h, x:x+w]

    # debug 이미지
    debug = original.copy()

    cv2.rectangle(
        debug,
        (x,y),
        (x+w, y+h),
        (0,255,0),
        3
    )

    return crop, debug


# =====================================
# 전체 이미지 처리
# =====================================

image_exts = (".jpg", ".jpeg", ".png", ".JPG")

files = [
    f for f in os.listdir(INPUT_DIR)
    if f.endswith(image_exts)
]

print(f"총 {len(files)}개 이미지 발견")

for fname in tqdm(files):

    path = os.path.join(INPUT_DIR, fname)

    img = cv2.imread(path)

    if img is None:
        continue

    cropped, debug = crop_leaf(img)

    cv2.imwrite(
        os.path.join(OUTPUT_DIR, fname),
        cropped
    )

    cv2.imwrite(
        os.path.join(DEBUG_DIR, fname),
        debug
    )

print("완료")
```

```python
from segment_anything import sam_model_registry
from segment_anything import SamAutomaticMaskGenerator

SAM_CHECKPOINT = "sam_vit_b_01ec64.pth"
MODEL_TYPE = "vit_b"

sam = sam_model_registry[MODEL_TYPE](
    checkpoint=SAM_CHECKPOINT
)

mask_generator = SamAutomaticMaskGenerator(
    sam,
    min_mask_region_area=1000
)

print("SAM 로드 완료")
```

```python
import os

INPUT_DIR = "./test"
OUTPUT_DIR = "./sam_crop"

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)

for fname in os.listdir(INPUT_DIR):

    if not fname.lower().endswith(
        (".jpg", ".jpeg", ".png")
    ):
        continue

    path = os.path.join(
        INPUT_DIR,
        fname
    )

    image = cv2.imread(path)

    image = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2RGB
    )

    masks = mask_generator.generate(image)

    if len(masks) == 0:
        continue

    largest_mask = max(
        masks,
        key=lambda x: x["area"]
    )

    mask = largest_mask["segmentation"]

    result = np.zeros_like(image)

    result[mask] = image[mask]

    ys, xs = np.where(mask)

    x1 = xs.min()
    x2 = xs.max()

    y1 = ys.min()
    y2 = ys.max()

    crop = result[y1:y2, x1:x2]

    cv2.imwrite(
        os.path.join(
            OUTPUT_DIR,
            fname
        ),
        cv2.cvtColor(
            crop,
            cv2.COLOR_RGB2BGR
        )
    )
```
